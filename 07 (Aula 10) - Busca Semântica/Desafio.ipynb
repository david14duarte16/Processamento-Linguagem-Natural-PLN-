# Desafio: Comparando Abordagens de Busca

Você deve criar um script de teste que compare o desempenho de duas abordagens de busca para uma consulta "traiçoeira".

* **A Query do Usuário:** "O aparelho descarrega muito rápido"
* **O Corpus de Teste:** Contém o review: "A autonomia da bateria é decepcionante"

```python
# Defina seu corpus de teste e a query
corpus_teste = [
    "A autonomia da bateria é decepcionante",
    "A tela do celular quebrou na primeira queda",
    "Ótimo custo benefício, entrega rápida"
]
query_usuario = ["O aparelho descarrega muito rápido"]

# ETAPA 1 (Busca por Frequência TF-IDF)
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

print("--- ETAPA 1: Busca por Frequência (TF-IDF) ---")
vectorizer = TfidfVectorizer()
matrix_corpus = vectorizer.fit_transform(corpus_teste)
vetor_query = vectorizer.transform(query_usuario)

# Calcula a similaridade de cosseno
similaridade_tfidf = cosine_similarity(vetor_query, matrix_corpus)[0]

for i, score in enumerate(similaridade_tfidf):
    print(f"Documento {i+1} | Score TF-IDF: {score:.4f} -> '{corpus_teste[i]}'")

# ETAPA 2 (Busca por Significado com spaCy Embeddings) e ETAPA 3 (Análise Comparativa)
import spacy

print("\n--- ETAPA 2: Busca por Significado (Embeddings spaCy) ---")
# Carregando o modelo em português
nlp = spacy.load("pt_core_news_md")

# Processando a query
doc_query = nlp(query_usuario[0])

# Calculando a similaridade
for i, texto in enumerate(corpus_teste):
    doc_corpus = nlp(texto)
    similaridade_spacy = doc_query.similarity(doc_corpus)
    print(f"Documento {i+1} | Score spaCy: {similaridade_spacy:.4f} -> '{texto}'")
```

## O que deve ser entregue (Insights de Especialista)

**1. Ortogonalidade:**
Na Etapa 1, os vetores eram ortogonais?
Sim. Como não há nenhuma palavra em comum entre a query e o review, o TF-IDF não encontra interseção. Matematicamente, os vetores formam 90 graus (ortogonais), resultando em produto escalar e similaridade de cosseno exatos a 0.0.

**2. Hipótese Distribucional:**
Como os Embeddings "sabem" que "descarrega rápido" e "autonomia decepcionante" habitam o mesmo bairro?
Porque o modelo aprende com o contexto. Palavras como "bateria" e "descarrega" aparecem perto das mesmas palavras vizinhas durante o treino. Isso faz com que seus vetores fiquem próximos no espaço, permitindo achar similaridade mesmo sem palavras exatas.

**3. Sugestão de Escalabilidade:**
Se tivesse 1 milhão de reviews, como resolveria o custo computacional?
Usando algoritmos de Approximate Nearest Neighbors (ANN) e bancos vetoriais como FAISS ou Pinecone. Eles agrupam os vetores parecidos e indexam os dados, reduzindo o tempo de busca para milissegundos, sem precisar comparar um por um.
